```python
def recommend_pattern(use_case: str) -> List[str]:
    text = use_case.lower()
    if "routing" in text or "specialist" in text:
        return ["Supervisor workflow", "Handoff orchestration"]
    if "parallel" in text or "independent" in text:
        return ["Parallel workflow", "Plan-based orchestration"]
    if "gate" in text or "validation" in text:
        return ["Conditional workflow", "Sequential workflow"]
    return ["Sequential workflow", "Plan-based orchestration"]
```
